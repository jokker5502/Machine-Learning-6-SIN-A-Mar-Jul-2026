## 1.3 Limpieza y preprocesamiento (0.2 pts)

Construyan el dataset listo para modelar. Como mínimo deben:

1. **Eliminar** features que no aportan información para clasificación (ej. `PassengerId`, `Name`, `Ticket`, `Cabin` — justifiquen cuáles eliminan).
2. **Imputar** valores nulos de manera defendible. ⚠️ **Cuidado con el data leakage**: la imputación debe aprenderse del training set, no del dataset completo.
3. **Codificar** variables categóricas (`Sex`, `Embarked`) usando one-hot encoding o equivalente.
4. **Escalar** las features numéricas con `StandardScaler`. ⚠️ Mismo cuidado: ajustar solo en training, transformar en test.

Pueden hacer la limpieza en este orden o usar un `Pipeline` de sklearn. Lo importante es **no contaminar el test set con información del training**.

```python
# 1.3 Limpieza — completen aquí

columnas_irrelevantes = ['PassengerId', 'Name', 'Ticket', 'Cabin']
df_filtrado = df.drop(columns=columnas_irrelevantes)

X = df_filtrado.drop(columns=['Survived'])
y = df_filtrado['Survived']

print("=== Estructura del Dataset Filtrado ===")
print(f"Dimensiones de X: {X.shape}")
print(f"Variables predictoras que quedan: {list(X.columns)}")

```

## 1.4 División train / validation / test (0.1 pts)

Dividan los datos en **3 conjuntos**: entrenamiento (60%), validación (20%) y test (20%). Usen `MI_SEMILLA` en todos los `random_state` y estratifiquen por `Survived` para mantener el balance.

> Recordatorio del bloque B de la última clase: el validation se usa para tunear hiperparámetros, el test se mira UNA SOLA VEZ al final.

```python
# 1.4 Split — completen aquí

X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y, 
    test_size=0.20, 
    random_state=MI_SEMILLA, 
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val, 
    test_size=0.25, 
    random_state=MI_SEMILLA, 
    stratify=y_train_val
)

print(" Dimensiones de los Conjuntos ")
print(f"Entrenamiento (60%): X_train = {X_train.shape}, y_train = {y_train.shape}")
print(f"Validación    (20%): X_val   = {X_val.shape}, y_val   = {y_val.shape}")
print(f"Test          (20%): X_test  = {X_test.shape}, y_test  = {y_test.shape}")

print("\n Verificación del Balance de Clases (Stratify) ")
print(f"Tasa de supervivencia en Train: {y_train.mean():.4f}")
print(f"Tasa de supervivencia en Val:   {y_val.mean():.4f}")
print(f"Tasa de supervivencia en Test:  {y_test.mean():.4f}")
```

Estratificación: Al pasar el parámetro stratify, nos aseguramos de que la tasa base de supervivencia (~38.38%) se mantenga idéntica en los tres conjuntos, evitando que un bloque se quede con más sobrevivientes que otro por puro azar.

Uso de Semilla: Se fija random_state=MI_SEMILLA (8854) en ambas divisiones para asegurar la reproducibilidad exacta del experimento cada vez que se ejecute el notebook.

## 1.5 Modelo baseline (0.2 pts)

Entrenen una `LogisticRegression` con **parámetros por defecto** (sin tunear nada) sobre el training set. Reporten:

- Accuracy en training
- Accuracy en validation  
- ¿Hay sobreajuste? Comenten brevemente.

Este baseline es la línea base contra la que se va a comparar el modelo tuneado.

```python
# 1.5 Baseline — completen aquí

X_train_clean = X_train.copy()
X_val_clean = X_val.copy()

# Imputación aprendida solo de Train 
num_imputer = SimpleImputer(strategy='median')
cat_imputer = SimpleImputer(strategy='most_frequent')

X_train_clean[['Age']] = num_imputer.fit_transform(X_train_clean[['Age']])
X_val_clean[['Age']] = num_imputer.transform(X_val_clean[['Age']])

X_train_clean[['Embarked']] = cat_imputer.fit_transform(X_train_clean[['Embarked']])
X_val_clean[['Embarked']] = cat_imputer.transform(X_val_clean[['Embarked']])

# Codificación Categórica 
X_train_encoded = pd.get_dummies(X_train_clean, columns=['Sex', 'Embarked'], drop_first=True)
X_val_encoded = pd.get_dummies(X_val_clean, columns=['Sex', 'Embarked'], drop_first=True)
X_train_encoded, X_val_encoded = X_train_encoded.align(X_val_encoded, join='left', axis=1, fill_value=0)

#  Escalado Numérico aprendido SOLO de Train
scaler = StandardScaler()
cols_a_escalar = ['Pclass', 'Age', 'SibSp', 'Parch', 'Fare']

X_train_scaled = X_train_encoded.copy()
X_val_scaled = X_val_encoded.copy()

X_train_scaled[cols_a_escalar] = scaler.fit_transform(X_train_scaled[cols_a_escalar])
X_val_scaled[cols_a_escalar] = scaler.transform(X_val_scaled[cols_a_escalar])

# Entrenamiento del Modelo Baseline
baseline_model = LogisticRegression(random_state=MI_SEMILLA)
baseline_model.fit(X_train_scaled, y_train)

# Accuracy
acc_train = baseline_model.score(X_train_scaled, y_train)
acc_val = baseline_model.score(X_val_scaled, y_val)

print(" Reporte del Modelo Baseline ")
print(f"Accuracy en Training:   {acc_train:.4f}")
print(f"Accuracy en Validation: {acc_val:.4f}")
print(f"Diferencia absoluta:    {abs(acc_train - acc_val):.4f}")
```

Análisis de Sobreajuste (Overfitting): Al evaluar los resultados con la semilla 8854, observamos que la precisión en el conjunto de entrenamiento y el de validación se mantienen muy cercanas entre sí (con una diferencia generalmente menor al 3% o 4%).

Conclusión: Debido a esta brecha tan estrecha, no hay evidencia de sobreajuste significativo. La Regresión Logística con los parámetros por defecto actúa como un modelo lineal simple que penaliza de forma intrínseca los cambios drásticos, logrando una capacidad de generalización inicial bastante aceptable sobre datos no vistos. Este comportamiento establece una línea base sólida para comparar con la optimización de hiperparámetros.

## 1.6 Modelo regularizado con búsqueda de hiperparámetros (0.3 pts)

Entrenen una `LogisticRegression` regularizada usando `GridSearchCV` con 5-fold cross-validation **sobre el training set** (NO sobre validation ni test). 

Como mínimo, exploren:
- `C`: al menos 5 valores en escala logarítmica entre 0.001 y 100
- `penalty`: al menos `['l1', 'l2']`

```python
# 1.6 GridSearchCV — completen aquí
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

# 1. Definición del espacio de búsqueda en escala logarítmica
param_grid = [
    {
        'C': [0.001, 0.01, 0.1, 1, 10, 100],
        'penalty': ['l1', 'l2'],
        'solver': ['liblinear']
    },
    {
        'C': [0.001, 0.01, 0.1, 1, 10, 100],
        'penalty': ['l1', 'l2'],
        'solver': ['saga']
    }
]

# 2. Configuración del GridSearchCV empleando validación cruzada de 5 pliegues
grid_search = GridSearchCV(
    estimator=LogisticRegression(random_state=MI_SEMILLA, max_iter=5000),
    param_grid=param_grid,
    cv=5,
    scoring='roc_auc',
    return_train_score=True
)

# 3. Ajustar el modelo solo para el conjunto de entrenamiento
grid_search.fit(X_train_scaled, y_train)

# 4. Extraer el mejor estimador obtenido
best_model = grid_search.best_estimator_

# 5. Evaluar el mejor modelo sobre el validation set usando probabilidades predictivas
y_val_probabilities = best_model.predict_proba(X_val_scaled)[:, 1]
validation_roc_auc = roc_auc_score(y_val, y_val_probabilities)

print(" Resultados de la Búsqueda en Malla (GridSearchCV) ")
print(f"1. Mejor combinación de hiperparámetros: {grid_search.best_params_}")
print(f"2. Mejor score de CV (ROC-AUC promedio):  {grid_search.best_score_:.4f}")
print(f"3. Score ROC-AUC sobre validation set:    {validation_roc_auc:.4f}")
```

Métrica elegida: Se seleccionó scoring='roc_auc' (Área bajo la curva ROC).

Justificación: La aseguradora TravelSafe necesita ajustar el precio de sus pólizas según el nivel de riesgo del cliente, lo que se traduce matemáticamente en evaluar probabilidades de supervivencia continuas, no predicciones binarias rígidas. ROC-AUC mide la capacidad del modelo para ordenar correctamente los perfiles de los pasajeros de mayor a menor riesgo a lo largo de todos los umbrales posibles, superando las limitaciones del accuracy por defecto, el cual se ciega en un único umbral fijo de 0.5.

### Gestión de Riesgo y Estrategia de Modelado

En **TravelSafe Insurance**, nuestro objetivo como aseguradora consolidada es proteger la solvencia y las reservas técnicas. En este negocio, un error de predicción tiene un costo asimétrico:

* **Falso Positivo (error crítico):** Predecir que un pasajero sobrevivirá cuando no será así. Esto nos lleva a cobrar una prima baja y asumir un siniestro catastrófico imprevisto, golpeando directamente nuestro patrimonio.
* **Falso Negativo (costo comercial):** Cobrar de más a quien no va a tener un siniestro. Aunque nos resta competitividad frente a *startups* agresivas, no pone en riesgo nuestra estabilidad financiera.

Por esta razón, nuestra prioridad absoluta es la **Precisión**. Para reflejar comercialmente este balance, adoptamos la métrica **$F_{0.5}$**, la cual pondera la precisión con el doble de importancia que el *recall*, penalizando severamente los falsos positivos.

### Optimización del Umbral (Threshold)

Al evaluar el set de validación con el umbral por defecto (0.5), obtuvimos una precisión de 67.57% y un recall de 73.53%. Para nuestro perfil de riesgo, aceptar que un 32.43% de las pólizas de bajo costo terminen en siniestro es inadmisible.

Analizando la curva Precision-Recall, decidimos **elevar el umbral a 0.58**. Este ajuste estratégico sacrifica algo de volumen comercial (*recall*) de manera controlada, pero eleva la precisión hacia un rango seguro del 75%-80%, blindando el capital de TravelSafe ante eventos de alta siniestralidad.

```python
# 1.8 Evaluación final única sobre el Test Set con umbral optimizado

X_test_clean = X_test.copy()

X_test_clean[['Age']] = num_imputer.transform(X_test_clean[['Age']])
X_test_clean[['Embarked']] = cat_imputer.transform(X_test_clean[['Embarked']])

X_test_encoded = pd.get_dummies(X_test_clean, columns=['Sex', 'Embarked'], drop_first=True)

X_train_columns = X_train_scaled.columns
for col in X_train_columns:
    if col not in X_test_encoded.columns:
        X_test_encoded[col] = 0

X_test_encoded = X_test_encoded[X_train_columns]

X_test_scaled = X_test_encoded.copy()
X_test_scaled[cols_a_escalar] = scaler.transform(X_test_scaled[cols_a_escalar])

y_test_proba = best_model.predict_proba(X_test_scaled)[:, 1]

NUEVO_UMBRAL = 0.58
y_test_pred_adjusted = (y_test_proba >= NUEVO_UMBRAL).astype(int)

test_accuracy = accuracy_score(y_test, y_test_pred_adjusted)
test_precision = precision_score(y_test, y_test_pred_adjusted)
test_recall = recall_score(y_test, y_test_pred_adjusted)
test_f1 = f1_score(y_test, y_test_pred_adjusted)
test_roc_auc = roc_auc_score(y_test, y_test_proba)

print(" EVALUACIÓN FINAL SOBRE TEST SET (UMBRAL: 0.58) ")
print(f"Accuracy final en Test:  {test_accuracy:.4f}")
print(f"Precision final en Test: {test_precision:.4f}")
print(f"Recall final en Test:    {test_recall:.4f}")
print(f"F1-Score final en Test:  {test_f1:.4f}")
print(f"ROC-AUC final en Test:   {test_roc_auc:.4f}")
```

El Bug: Se está ejecutando el escalado (scaler.fit_transform(X)) sobre la totalidad del dataset antes de realizar la separación de los conjuntos de entrenamiento y prueba (train_test_split).

Por qué es un error: Provoca un fenómeno crítico llamado Data Leakage (Fuga de datos). Al calcular la media y la desviación estándar global del dataset completo, la información y distribución del conjunto de prueba (Test Set) se filtra directamente dentro del conjunto de entrenamiento.

Impacto en producción: En la práctica, el modelo "hace trampa" porque entrena conociendo sutilmente las métricas del test set. Esto se traduce en métricas de evaluación falsamente optimistas durante el desarrollo, pero un rendimiento deficiente e inestable cuando el modelo se despliega en producción y se enfrenta a datos reales completamente nuevos.
